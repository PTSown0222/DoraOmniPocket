# 1. From tiktoken to scratch

```python
import tiktoken
```

```python
gpt2_tokenizer = tiktoken.get_encoding("gpt2")
gpt2_tokenizer.encode('get MORE texxt :v')
```

`GPT-2 tokenizer nhận vào một list các token IDs và trả về string tương ứng`

```python
for i in range(10):
    decoded = gpt2_tokenizer.decode([i])
    print(f"{i} -> {decoded}")
```
